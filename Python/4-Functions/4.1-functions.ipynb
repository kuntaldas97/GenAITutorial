```python
def even_or_odd(num):
    """This function performs even or odd"""
    if num%2==0:
        print("The number is even")
    else:
        print("The number is odd")
```

```python
### Variable length Aeguments
## Positional and Keywords arguments

def print_numbers(*krish):
    for number in krish:
        print(number)
        
```

```python
print_numbers(1,2,3,4,5,6,7,"Kuntal") 
```

```python
### Positional Arguments
## While calling the function when all the parameters are not in key-value pairs, we can make use of positional arguments (*args)
def print_numbers(*args):
  for item in args:
    print(item)
```

```python
print_numbers(1,2,3,4,5,6,7,8,"kuntal")
```

```python
### Keywords Arguments
## while calling the function when all the parameters are in key-value pairs then we can use keywords arguments(**kwargs)
def print_details(**kwargs):
  for key,value in kwargs.items():
    print(f"{key}:{value}")
```

```python
print_details(name="Kuntal", age="28", country="India")
```

```python
def print_details(*args,**kwargs):
  for val in args:
    print(f"Positional argument : {val}")
  for key,value in kwargs.items():
    print(f"{key}:{value}")
```

```python
print_details(1,2,3,name="Kuntal", age="28", country="India")
```
