Example 1 : Temperature Conversion

```python
def convert_temperature(temp,unit):
  """This function converts temperature between Celcius and Fahrenhait"""
  if unit=='C':
    return (temp*9)/5+32
  elif unit=='F':
    return (temp-32)*5/9
  else:
    return None
ans = convert_temperature(37,'C')
print(ans)
```

Example 2 : Password Strength Checker


```python
def is_strong_password(password):
  """This function checks if the password is strong or not"""
  if len(password)<8:
    return False
  if not any(char.isdigit() for char in password):
    return False
  if not any(char.isupper() for char in password):
    return False
  if not any(char.islower() for char in password):
    return False
  if not any(char in  "[!\"#$%&'()*+,\\-./:;<=>?@[\\\\\\]^_`{|}~]" for char in password):
    return False
  return True

## calling the function
print(is_strong_password("weekPswd"))
print(is_strong_password('Str0@gPswD'))
```

Example 3 : Calculate the Total Cost Of Items In a Shopping Cart

```python
## defining the function
def calculate_total_cost(cart):
  totalCost=0
  for item in cart:
    totalCost += item['price']*item['quantity']
  return totalCost

## Details of the cart data
cart = [
    {'name':'Apple', 'price':5, 'quantity':4},
    {'name':'Banana', 'price':3, 'quantity':6},
    {'name':'Orrange', 'price':7, 'quantity':3}
]
##Calling and printing the function
print(calculate_total_cost(cart))
```

Example 4 : Check if a String is Palindrome

```python
def is_palindrome(s):
  s=s.lower().replace(" ","")
  return s==s[::-1]

print(is_palindrome("racecar"))
print(is_palindrome("Kuntal"))
```

Example 5 : Calculate the factorial of a number using recursion

```python
def factorial(n):
  if n==0:
    return 1
  else:
    return n*factorial(n-1)

print("Enter the number : ")
num = int(input())
print(f"Factorial of is {num} is {factorial(num)}")
```

Example 6 : A function to read a file and count the frequeny of each worda

```python
def count_word_frequency(file_path):
  word_count={}
  with open(file_path,'r') as file:
    for line in file:
        words=line.split()
        for word in words:
            word=word.lower().strip('.,!?;"\'')
            word_count[word]=word_count.get(word,0)+1
            
    return word_count

filepath='sample.txt'
word_frequency=count_word_frequency(filepath)
print(word_frequency)
```

Example 7 : Validate email address

```python
import re
# Email validation function
def is_valid_email(email):
    """This function checks if the email is valid or not"""
    pattern = r'^[A-Za-z0-9._%+-]+@[A-Za-z0-9.-]+\.[A-Za-z]{2,}$'
    return re.match(pattern,email) is not None

# Calling the function
print(is_valid_email("kuntal@example.com"))
print(is_valid_email("invalid-email"))
```
