```python
## Tuples are immutable, meaning their elements can not be modified or changed once assinged

tpl=(1,2,3,4)
print(tpl)

tpl[1]="Bittu" ## Will throw an error coz tuple can't be modified
```

```python
## Tuple methods

print(numbers.count(1))
print(numbers.index(3))
```

```python
## Packing and Unpacking tuple 
## packing
packed_tuple=1,"Helloworld",3.14
print(packed_tuple)
## unpacking a tuple
a,b,c = packed_tuple
print(a)
print(b)
print(c)
```

```python
## Unpacking with * 
numbers=(1,2,3,4,5,6)
first, *middle, last = numbers
print(first)
print(middle)
print(last)
```

```python
##Nested Tuple

lst = [[1,2,3,4],[6,7,8,9],[1,"Hello",3.14,'c']]
#getting the 3rd element of the first index
lst[0][2]

nested_tuple=((1,2,3),('a','b','c'), (True, False))
print(nested_tuple[0])
print(nested_tuple[1][:2])
```

```python
## iterating over nested tuples
for sub_tuple in nested_tuple:
    for item in sub_tuple:
        print(item,end=" ")
    print()
```
