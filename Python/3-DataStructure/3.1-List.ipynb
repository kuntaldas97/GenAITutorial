```python
### Accessing list elements

fruits=["apple", "banana", "cherry", "kiwi", "mango"]

print(fruits[0], fruits[-1], fruits[4])
print(fruits[1:])
print(fruits[1:3])
print(fruits[-1:])
print(fruits[-1:-2]) # there is no such index as -2 thus it will return []

fruits[1]="watermelon"
print(fruits)
#fruits[1:]="watermelon"
#print(fruits)

## List methods

fruits.append("orrange") #insert at the end
print(fruits)

fruits.insert(1,"banana") #insert at specific index
print(fruits)
fruits.insert(1,"banana") #insert at specific index
print(fruits)
fruits.remove("banana") #remove the first occurance of an item
print(fruits)

###remove and return the last element
popped_fruits = fruits.pop()
print(popped_fruits)
print(fruits)

## getting the index of a particular item
index=fruits.index("cherry")
print(index)

## counting of particular item
fruits.insert(2,"banana")
print(fruits,"\n",fruits.count("banana"))

## sorting the list in ascending order
fruits.sort()
print(fruits)

## reversing the list
fruits.reverse()
print(fruits)

## clear the entire list
fruits.clear()
print(fruits)
```

## List Comprehension

Basic Syntax      [expression for item in iterable]

with conditional logic   [expression for item in iterable if condition] 

Nested List Comprehension    [expression for item1 in iterable1 for item2 in itereable2]

```python
###Basic list comprehension
square=[num**2 for num in range(10)]
print(square)
```

```python
# List comeprehension with Condition
lst=[]
for i in range(10):
    if i%2==0:
        lst.append(i)
print(lst)

## for the above POC I'll apply list comprehension
even_numbers=[num for num in range(10) if num%2==0]
print(even_numbers)
```

```python
### Nested List Comprehension

lst1=[1,2,3,4]
lst2=['a','b','c','d']

pair=[[i,j] for i in lst1 for j in lst2]
print(pair)
```

```python
## List comprehension with function calls
words = ["Hello", "World", "Python", "List", "Comprehension"]
lengths=[len(word) for word in words]
print(lengths)
```

```python
## for both is and else condition

nums = [1,2,3,4,5]
lebels = ["Even" if num%2==0 else "Odd" for num in nums]
print(lebels)
```
