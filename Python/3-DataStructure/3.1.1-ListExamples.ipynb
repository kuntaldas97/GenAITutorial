```python
###Ctreting a to-do-list

to_do_list=["Buy Groceries","Clean the house","Pay bills"]

## Adding to task
to_do_list.append("Schedule meeting")
to_do_list.append("Go for a run")

## removing a complete task
to_do_list.remove("Clean the house")

## checking if a task is in the list

if "Pay bills" in to_do_list:
    print("Don't forget to pay the utility bills")
print("Remaining to-do-list")
for task in to_do_list:
    print(f"-{task}")

```

Managing an inventry


```python
## Managing an inventory
inventory = ["apples","bananas","oranges","grapes"]

#Adding a new item
inventory.append("strawbarries")

#removing an item that is out of the stock
inventory.remove("bananas")

#Checking if an item is in the stock or not
item="oranges"
if item in inventory:
    print(f"{item} are in stock.")
else:
    print(f"{item} are out of stock.")

#Printing the inventory list
for item in inventory:
    print(f"-{item}")
```

Create a List for customer feedback


```python
## creating the list
feedback = ["Great service", "Very stisfied","Could be better","Excelent experient"]
positiveFeedbackCount = sum(1 for comment in feedback if "great" in comment.lower() or
"excelent" in comment.lower())
print(f"Positive Feedback Count: {positiveFeedbackCount}")

# Printing all feedback
print("User Feedback :")
for comment in feedback:
    print(f"-{comment}")
```
